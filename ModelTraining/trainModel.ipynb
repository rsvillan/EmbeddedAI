```python
import socket
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import numpy as np
import matplotlib.pyplot as plt
import time
```

```python
# --- CIFAR-10 Classes ---
cifar10_classes = [
    "Airplane", "Automobile", "Bird", "Cat", "Deer", 
    "Dog", "Frog", "Horse", "Ship", "Truck"
]
```

```python
class HardwareCNN(nn.Module):
    def __init__(self):
        super(HardwareCNN, self).__init__()
        self.conv1 = nn.Conv2d(1, 16, kernel_size=3, padding=1)

        self.conv2 = nn.Conv2d(16, 32, kernel_size=3, padding=1)

        self.pool = nn.MaxPool2d(2, 2) # 32 -> 16 -> 8
        
        self.fc1 = nn.Linear(32 * 8 * 8, 10)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.pool(self.relu(self.conv2(x)))

        x = x.view(-1, 32 * 8 * 8)
        
        x = self.fc1(x)
        return x
```

```python
transform = transforms.Compose([
        transforms.Resize((32, 32)),
        transforms.Grayscale(num_output_channels=1),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))])

trainset = torchvision.datasets.CIFAR10(root='../data', train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=32, shuffle=True) 
```

```python
img_tensor, label_idx = trainset[9]

img_denorm = img_tensor * 0.5 + 0.5

img_np = img_denorm.squeeze(0).numpy()

plt.imshow(img_np, cmap='gray')
plt.title(f"Label: {cifar10_classes[label_idx]} (Class {label_idx})")
plt.axis('off')
plt.show()
```

```python
model = HardwareCNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("Training hardware-accurate model...")
model.train()

epochs = 10
for epoch in range(epochs):
    running_loss = 0.0
    for i, data in enumerate(trainloader, 0):
        inputs, labels = data
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        if i % 200 == 199:
            print(f"Epoch {epoch+1} | Batch {i+1} | Loss: {running_loss / 200:.3f}")
            running_loss = 0.0

print("Finished Training. Saving weights...")
torch.save(model.state_dict(), "model_weights.pth")
```
